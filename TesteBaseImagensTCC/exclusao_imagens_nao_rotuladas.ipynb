```python
from pathlib import Path

pasta_dirty = Path.cwd() / "Lixo - Noite"
pasta_labels = pasta_dirty / "labels_2"
extensoes_imagem = {".jpg", ".jpeg", ".png", ".webp", ".bmp", ".tif", ".tiff", ".avif"}

if not pasta_dirty.is_dir():
    raise FileNotFoundError(f"Pasta de imagens não encontrada: {pasta_dirty}")
if not pasta_labels.is_dir():
    raise FileNotFoundError(f"Pasta de rótulos não encontrada: {pasta_labels}")

nomes_rotulados = {arquivo.stem.casefold() for arquivo in pasta_labels.glob("*.json") if arquivo.is_file()}
imagens_sem_rotulo = [
    arquivo
    for arquivo in pasta_dirty.iterdir()
    if arquivo.is_file()
    and arquivo.suffix.casefold() in extensoes_imagem
    and arquivo.stem.casefold() not in nomes_rotulados
]

for imagem in imagens_sem_rotulo:
    imagem.unlink()

print(f"Imagens excluídas por não terem JSON correspondente: {len(imagens_sem_rotulo)}")
print(f"Imagens mantidas: {sum(1 for arquivo in pasta_dirty.iterdir() if arquivo.is_file() and arquivo.suffix.casefold() in extensoes_imagem)}")
```
